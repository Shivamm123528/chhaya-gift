# Prototype: test content and layout before running Streamlit
Run from the repo folder.

```python
import os, random
import gift_data as g
print(g.SPLASH_TITLE)
print(len(g.ENVELOPES), 'envelopes,', len(g.MEMORIES), 'memories')
```

## 1. Content checks
Paragraph length (should be short) and leftover placeholders.

```python
def all_text():
    t = list(g.SPLASH_LINE) + g.CLOSING + g.GRATEFUL + g.LOVE_LIST + g.SMILE_MEMORIES + g.NEXT_UP
    for e in g.ENVELOPES: t += e['lines']
    for n in g.VOICE_NOTES: t.append(n['script'])
    t += [m[1] for m in g.MEMORIES] + [w[1] for w in g.WISHES]
    return t
for s in all_text():
    if len(s.split()) > 60: print('LONG:', s[:60], '...')
print('Placeholders left:', sum('[' in s for s in all_text()))
```

## 2. Asset check
What will show and what will be hidden.

```python
files = ['assets/background.jpeg'] + g.PHOTOS + [n['file'] for n in g.VOICE_NOTES]
for f in files: print('OK     ' if os.path.exists(f) else 'hidden ', f)
```

## 3. Palette preview

```python
import matplotlib.pyplot as plt
pal = {'blush':'#f8d7e0','lavender':'#e3d9f5','sky':'#d4e9f7','cream':'#fff8f1','ink':'#4b3f58'}
fig, ax = plt.subplots(figsize=(7,1.2))
for i,(k,v) in enumerate(pal.items()):
    ax.add_patch(plt.Rectangle((i,0),1,1,color=v)); ax.text(i+.5,.5,k,ha='center',color='#4b3f58' if k!='ink' else 'w')
ax.set_xlim(0,len(pal)); ax.set_ylim(0,1); ax.axis('off'); plt.show()
```

## 4. Layout simulation
Two-column memory cards, as in the app.

```python
for i in range(0, len(g.MEMORIES), 2):
    row = g.MEMORIES[i:i+2]
    print(' | '.join(f'{t:<28}' for t,_ in row))
```
